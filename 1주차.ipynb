## 4. 조건 필터링 (행 선택) - 데이터 분석의 핵심 기술
-조건 필터링은 필요한 데이터만 골라서 보는 것

- df[조건] 

- 비교연산자 : >, >=, <, <=, ==, !=

- 예: 평균기온이 20도 이상인 데이터를 선택,  df[df['avgTa'] >= 20]

```python
df['avgTa']  > 20     # 조건
```

```python
df['avgTa']  > 20
```

```python
df[ df['avgTa']  > 20   ]
```

```python
df['avgTa']  > 20     # 조건
```

### 4-2. 두 개 이상의 조건

- &  : and 연산자
- |  : or 연산자
- 주의: 각각의 조건을 괄호 (   ) 로 묶어줄것

항상 조건을 먼저 만들어 봅니다

문: 최저 기온이 0 이상이고 강수량이 5 이하인 데이터를 선택

```python
df['minTa'] >= 0
```

```python
df['minTa'] >= 0
```

```python
(df['minTa'] >= 0) & (df['sumRn'] <= 5)
```

```python
(df['minTa'] >= 0) & (df['sumRn'] <= 5)
```

```python
df[ (df['minTa'] >= 0) & (df['sumRn'] <= 5) ]
```

```python
df[ (df['minTa'] >= 0) & (df['sumRn'] <= 5) ]
```
