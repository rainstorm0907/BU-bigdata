# 2주차 데이터 탐색과 시각화 실습

pandas를 사용하여 데이터를 불러오고 데이터의 구조를 이해한 후, 데이터의 분포를 분석하고 matplotlib.pyplot 사용하여 시각화하는 방법을 학습한다.

## 2주차 학습 목표

1. CSV 데이터를 pandas로 불러올 수 있다.  
   - `pd.read_csv()`를 사용하여 CSV 파일을 데이터프레임으로 읽는다.

2. 데이터프레임의 구조를 탐색할 수 있다.  
   - `info()`를 사용하여 데이터 타입과 결측값 여부를 확인한다.  
   - `columns`를 사용하여 컬럼 이름을 확인한다.  
   - `describe()`를 사용하여 수치 데이터의 기본 통계를 확인한다.

3. 특정 컬럼 데이터를 선택할 수 있다.  
   - `df['컬럼명']`을 사용하여 특정 컬럼을 선택한다.

4. 데이터의 고유값을 확인할 수 있다.  
   - `unique()`를 사용하여 해당 컬럼에 어떤 값들이 존재하는지 확인한다.

5. 데이터의 빈도 분포를 분석할 수 있다.  
   - `value_counts()`를 사용하여 값의 등장 횟수를 계산한다.

6. 데이터 분석 결과를 시각화할 수 있다.  
   - `value_counts().plot()`를 사용하여 값의 출현 횟수를 그래프로 시각화한다.
   - `value_counts().plot(kind='bar')`   또는 `value_counts().plot.bar()`를 사용하여 값의 출현 횟수를 막대그래프로 시각화한다.
   - 를 사용하여 값의 출현 횟수를 막대그래프로 시각화한다
   
7. `matplotlib.pyplot` 을 사용하여 그래프를 설정할 수 있다.  
   - `plt.title()`을 사용하여 그래프 제목을 설정한다.  
   - `plt.xlabel()`과 `plt.ylabel()`을 사용하여 축 이름을 설정한다.  
   - `plt.show()`를 사용하여 그래프를 화면에 출력한다.

8. `groupby()`를 사용하여 데이터를 그룹별로 분석할 수 있다.  
   - `groupby()`를 사용하여 특정 기준으로 데이터를 그룹화한다.

9. 그룹별 통계값을 계산하여 데이터의 특징을 분석할 수 있다.  
   - `groupby()`와 집계 함수를 사용하여 평균, 최소, 최대, 합계, 갯수 등의 통계를 계산한다.

### unique() 와 value_counts() 의 차이

데이터를 분석할 때 특정 컬럼에 어떤 값들이 존재하는지 확인하는 것이 중요  
pandas에서는 이를 위해 `unique()`와 `value_counts()` 함수를 사용할 수 있다.

- `unique()`는 해당 컬럼에 존재하는 **고유한 값의 목록**을 반환  

- `value_counts()`는 해당 컬럼에 **각 값이 몇 번 등장하는지**를 반환하고 그 결과는 pandas Series 형태이므로 바로 그래프로 시각화할 수 있다.

```python
df['stnNm'].unique()
```

## 그룹핑 : 묶어서 계산하기

- groupby() 함수 : 기준 컬럼으로 데이터를 묶은 뒤  묶음별 통계값을 계산



### 문제 4) 지역 ('stnNm') 별 평균기온('avgTa') 의 평균을 계산

```python
df.groupby('stnNm')['avgTa'].mean()
```

```python
df.groupby('stnNm')['avgTa'].mean()
```

```python
type(df.groupby('stnNm')['avgTa'].mean())
```

```python
df.groupby('stnNm')['avgTa'].mean().plot(kind='bar')
plt.title('지역별 연간 평균기온')
plt.xlabel('지역')
plt.ylabel('평균기온')
plt.xticks(rotation=45)
plt.show()
```

```python
df.groupby('stnNm')['avgTa'].mean().plot(kind='bar')
plt.title('지역별 연간 평균기온')
plt.xlabel('지역')
plt.ylabel('평균기온')
plt.xticks(rotation=45)
plt.show()
```

### 문제 5) 달 ('month') 별 강수량 ('sumRn') 의 평균을 계산

```python
df.groupby('month')['sumRn'].mean()
```

```python
df.groupby('month')['sumRn'].mean()
```

```python
df.groupby('month')['sumRn'].mean().plot(marker='o')
plt.title('월별 평균 강수량')
plt.xlabel('월')
plt.ylabel('강수량')
plt.xticks(df.groupby('month')['sumRn'].mean().index)
plt.show()
```

```python
df.groupby('month')['sumRn'].mean().plot(marker='o')
plt.title('월별 평균 강수량')
plt.xlabel('월')
plt.ylabel('강수량')
plt.xticks(df.groupby('month')['sumRn'].mean().index)
plt.show()
```
